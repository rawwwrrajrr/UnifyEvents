```python

import heapq

def h(state, goal):
    return sum(1 for i in range(9) if state[i] != 0 and state[i] != goal[i])

def astar(start, goal):
    q = [(h(start, goal), 0, start, [start])]
    visited = {}

    while q:
        f, g, state, path = heapq.heappop(q)

        if state == goal:
            print("Goal reached!")
            print("Total moves:", g)
            for s in path:
                print(s[:3])
                print(s[3:6])
                print(s[6:])
                print()
            return

        if state in visited and visited[state] <= g:
            continue
        visited[state] = g

        i = state.index(0)
        for j in [i-3, i+3, i-1, i+1]:
            if 0 <= j < 9 and abs(i//3 - j//3) + abs(i%3 - j%3) == 1:
                s = list(state)
                s[i], s[j] = s[j], s[i]
                s = tuple(s)
                heapq.heappush(q, (g+1+h(s, goal), g+1, s, path+[s]))

start = tuple(map(int, input("Enter initial state: ").split()))
goal = tuple(map(int, input("Enter goal state: ").split()))
astar(start, goal)
```
