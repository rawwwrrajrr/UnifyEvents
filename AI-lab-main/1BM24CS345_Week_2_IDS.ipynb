```python

def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    row, col = blank // 3, blank % 3

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            new_state = list(state)
            new_state[blank], new_state[pos] = new_state[pos], new_state[blank]
            neighbors.append(tuple(new_state))

    return neighbors


def display(state):
    for i in range(0, 9, 3):
        print(*state[i:i + 3])
    print()


def dls(state, goal, limit, path):
    if state == goal:
        return path

    if len(path) - 1 == limit:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in path:
            result = dls(neighbor, goal, limit, path + [neighbor])

            if result:
                return result

    return None


def ids(start, goal):
    for limit in range(32):
        print("Depth limit:", limit)
        result = dls(start, goal, limit, [start])

        if result:
            print("Goal found!")
            print("Number of moves:", len(result) - 1)

            for i, state in enumerate(result):
                print("Step", i)
                display(state)

            return

    print("Goal not found.")


def valid(state):
    return len(state) == 9 and set(state) == set(range(9))


start = tuple(map(int, input("Enter initial state (0 for blank): ").split()))
goal = tuple(map(int, input("Enter goal state (0 for blank): ").split()))

if valid(start) and valid(goal):
    ids(start, goal)
else:
    print("Invalid input! Enter numbers 0 to 8 exactly once.")
```
